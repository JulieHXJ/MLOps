```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("../data/processed/final_enriched_rental_data.csv")
data["hour"] = pd.to_datetime(data["hour"])
data["date"] = pd.to_datetime(data["date"])
```

```python
hourly_data = (
    data.groupby("hour", as_index=False)
    .agg(
        system_total_count=("total_count", "sum"),
        system_registered_count=("registered_count", "sum"),
        system_direct_count=("direct_count", "sum"),
        zero_demand_locations=("is_zero_filled_rental", "sum"),
        hour_of_day=("hour_of_day", "first"),
        day_of_week=("day_of_week", "first"),
        month=("month", "first"),
        # quarter=("quarter", "first"),
        is_weekend=("is_weekend", "first"),
        is_workday=("is_workday", "first"),
        is_holiday=("is_holiday", "first"),
        conditions=("conditions", "first"),
        temperature_c=("temperature_c", "first"),
        perceived_temperature_c=("perceived_temperature_c", "first"),
        humidity=("humidity", "first"),
        windspeed_kmh=("windspeed_kmh", "first"),
        is_missing_weather=("is_missing_weather", "first"),
    )
)

hourly_data.head()
```
